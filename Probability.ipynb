```python
# ==========================================================
# รวมโค้ดคำนวณ Assignment 1 - 4
# ==========================================================

print("=== ASSIGNMENT 1 ===")
# 1.1 Marginalization
p_in_stock = 0.10 + 0.20 + 0.30
p_sold_out = 0.25 + 0.10 + 0.05
p_fall = 0.10 + 0.25
p_spring = 0.20 + 0.10
p_summer = 0.30 + 0.05
print(f"P(in_stock) = {p_in_stock:.2f}")
print(f"P(sold_out) = {p_sold_out:.2f}")
print(f"P(fall) = {p_fall:.2f}, P(spring) = {p_spring:.2f}, P(summer) = {p_summer:.2f}")

# 1.2 P(sold_out | fall)
p_sold_given_fall = 0.25 / 0.35
print(f"P(sold_out | fall) = {p_sold_given_fall:.4f}")

# 1.3 P(spring)
print(f"P(spring) = {p_spring:.2f}")

# 1.4 Normalization table P(Cookies | fall)
norm_in_stock = 0.10 / 0.35
norm_sold_out = 0.25 / 0.35
print(f"Normalized P(in_stock | fall) = {norm_in_stock:.4f}")
print(f"Normalized P(sold_out | fall) = {norm_sold_out:.4f}")

print("\n=== ASSIGNMENT 2 ===")
p_permanent = 0.35 + 0.10 + 0.05 + 0.15
p_haircut_and_permanent = 0.35 + 0.10
p_haircut_given_permanent = p_haircut_and_permanent / p_permanent
print(f"P(haircut | permanent) = {p_haircut_given_permanent:.4f}")

print("\n=== ASSIGNMENT 3 ===")
p_bike = 0.14
p_guard = 0.48
p_bike_and_guard = 0.12
p_bike_given_guard = p_bike_and_guard / p_guard
print(f"P(rides bike | crossing guard) = {p_bike_given_guard:.4f}")

print("\n=== ASSIGNMENT 4 ===")
p_on_315_given_crash = 0.8
p_crash = 0.1
p_on_315 = 0.7
p_crash_given_on_315 = (p_on_315_given_crash * p_crash) / p_on_315
print(f"P(crash | on_315) = {p_crash_given_on_315:.4f}")
```
