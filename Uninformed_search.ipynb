 1.2เขียนโค้ดแสดงเส้นทางในรูปแบบต่างๆ (tree traversal)

```python
from collections import deque

# โครงสร้างกราฟ (Adjacency List)
graph = {
    'A': [('D', 81), ('B', 213), ('I', 163)],
    'D': [('E', 76)],
    'E': [('J', 246)],
    'J': [('G', 80)],
    'B': [('C', 182), ('H', 99)],
    'C': [],
    'H': [('F', 179), ('G', 163)],
    'F': [],
    'I': [('G', 498)],
    'G': []
}

# 1. BFS Traversal
def bfs_traversal(start_node):
    visited_order = []
    queue = deque([start_node])

    while queue:
        node = queue.popleft()
        if node not in visited_order:
            visited_order.append(node)
            for neighbor, _ in graph[node]:
                queue.append(neighbor)
    return visited_order

# 2. DFS Traversal
def dfs_traversal(start_node, visited_order=None):
    if visited_order is None:
        visited_order = []

    visited_order.append(start_node)

    for neighbor, _ in graph[start_node]:
        if neighbor not in visited_order:
            dfs_traversal(neighbor, visited_order)

    return visited_order

# แสดงผลลัพธ์ข้อ 1.2
print("=== 1.2 Tree Traversal Results ===")
print("BFS Traversal Order :", " -> ".join(bfs_traversal('A')))
print("DFS Traversal Order :", " -> ".join(dfs_traversal('A')))
```

1.3เขียนโค้ดเพื่อแสดงคำตอบในรูปแบบต่างๆ (solution)

```python
from collections import deque
import heapq

# โครงสร้างกราฟ (Adjacency List)
graph = {
    'A': [('D', 81), ('B', 213), ('I', 163)],
    'D': [('E', 76)],
    'E': [('J', 246)],
    'J': [('G', 80)],
    'B': [('C', 182), ('H', 99)],
    'C': [],
    'H': [('F', 179), ('G', 163)],
    'F': [],
    'I': [('G', 498)],
    'G': []
}

# 1. BFS Solution (Shortest Path by Edges)
def bfs_solution(start, goal):
    queue = deque([(start, [start], 0)])

    while queue:
        (node, path, cost) = queue.popleft()
        if node == goal:
            return path, cost
        for neighbor, weight in graph[node]:
            queue.append((neighbor, path + [neighbor], cost + weight))
    return None

# 2. DFS Solution (Deepest Path First)
def dfs_solution(start, goal, path=None, cost=0):
    if path is None:
        path = [start]
    if start == goal:
        return path, cost

    for neighbor, weight in graph[start]:
        if neighbor not in path:
            res = dfs_solution(neighbor, goal, path + [neighbor], cost + weight)
            if res:
                return res
    return None

# 3. UCS Solution (Optimal Path by Lowest Cost)
def ucs_solution(start, goal):
    pq = [(0, start, [start])]
    visited = set()

    while pq:
        (cost, node, path) = heapq.heappop(pq)
        if node == goal:
            return path, cost
        if node not in visited:
            visited.add(node)
            for neighbor, weight in graph[node]:
                if neighbor not in visited:
                    heapq.heappush(pq, (cost + weight, neighbor, path + [neighbor]))
    return None

# แสดงผลลัพธ์ข้อ 1.3
bfs_path, bfs_cost = bfs_solution('A', 'G')
dfs_path, dfs_cost = dfs_solution('A', 'G')
ucs_path, ucs_cost = ucs_solution('A', 'G')

print("=== 1.3 Search Solution Results ===")
print(f"1. BFS Solution Path : {' -> '.join(bfs_path)} | Total Distance: {bfs_cost} km")
print(f"2. DFS Solution Path : {' -> '.join(dfs_path)} | Total Distance: {dfs_cost} km")
print(f"3. UCS Solution Path : {' -> '.join(ucs_path)} | Total Distance: {ucs_cost} km (Optimal)")
```
